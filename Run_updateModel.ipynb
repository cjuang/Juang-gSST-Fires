```python
# generate the sequence for adding uncertainty to the model.
# this sequence is saved and used throughout the model run.
# only update this if you want to change the 500-member ensemble sequence.

import numpy as np
from customconfig import *
predict_string = 'predicted//'+climindname+'//'

yearxaxis = np.arange(firstyear, finalyear+1)
iyears = np.arange(0,len(yearxaxis))

## first make matrix of random years (n_years x n_ensemble members)
# overwrite previous matrix of random years
# RULE: the randomized sequence of years MUST be the same for every ecoprovince to account for covariance.
istop = 0
errormatrix = np.empty((0,len(yearxaxis)))
while istop < (nensemble):
    tmprandom = np.array([np.random.choice(np.delete(iyears, i)) for i in range(len(iyears))])
    # randomly select errors and construct a timeseries
    tmprandom = tmprandom.reshape(1, -1)
    errormatrix = np.concatenate([errormatrix, 
                                  tmprandom], axis=0)
    istop+=1

# export error matrix so that it can be used again if needed
tmp_fname = predict_string + "Model_ENSOFull_ErrorSequence.txt"
np.savetxt(tmp_fname, errormatrix)

from datetime import datetime
now = datetime.now()
# dd/mm/YY H:M:S
dt_string = now.strftime("%d/%m/%Y %H:%M:%S")
print("ERRORMATRIX written to file =", dt_string)
```

**Update the models by running the notebooks in succession.**

* Note that this does **not** update the data.
* The SST index used will be taken from `0_climindname.txt` and applied in every notebook. The outputs will also go in the folder of the same name.

Possible climindnames (climate indices names)
* patch125-155_nino3-34
* pdo_noaa
* soi_noaa
* mei_noaa
* patch125-155_nino3-34a
* patch125-155_nino3-34b
* patch125-155_nino3-34c
* patch125-155_nino3-34d

```python
import papermill as pm

# Read the climate index from the .txt file
with open('0_climindname.txt', 'r') as f:
    climindname = f.read().strip()
print(climindname +' will be used for the SST gradient in the files')
```

```python
# execute notebooks to run the models
notebooks = [
    "Model_ENSOclim_CoeffCheck.ipynb",
    "Model_ENSOclim_CoeffCheck-avg.ipynb",
    "Model_ENSOclim_AkaikeCoeff.ipynb",
    "Model_Akaike.ipynb", # train model
    "Model_ENSOfull12-2022.ipynb", # climate-BA predictions
    "Model_ENSOfullclim12-2025.ipynb", # counterfactual climate
    "Model_ENSOpredictclim.ipynb",
    "Model_Akaike_CrossValid.ipynb",
]

for notebook in notebooks:
    print(f"Running {notebook}...")
    pm.execute_notebook(notebook, notebook)
    print(f"{notebook} completed!\n")
```
